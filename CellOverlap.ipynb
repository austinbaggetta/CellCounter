### Calculate overlap for multiple different markers.

```python
import os
import pandas as pd
import numpy as np
import json
import plotly.graph_objects as go
from os.path import join as pjoin

from preprocessing import (
    load_detected_cells,
    get_overlapping_cells,
    get_overlap_proportions
)

from visualizations import (
    fig_temp,
    total_cell_number
)
```

```python
## Settings
experimenter = 'Destynie'
mouse = 'm01'
cell_path = f'./demo_data/{experimenter}/output/{mouse}' ## all detected cell csvs should be in the same folder
output_path = f'./demo_data/{experimenter}/cell_overlap/{mouse}'

if not os.path.exists(output_path):
    os.makedirs(output_path)
```

```python
## Load your detected cell files into one dataframe
cell_df = load_detected_cells(cell_path, file_pattern='detected_cells.csv')
markers = cell_df['marker'].unique()
```

```python
## Plot total number of cells for each marker
num_cells = cell_df.groupby(['marker'], as_index=False).agg({'seed': 'count'})
fig = total_cell_number(num_cells)
fig.show()
```

```python
## Get overlapping cells
dist_thresh = 15 ## in pixels
denom_marker, overlap_df = get_overlapping_cells(cell_df, num_cells, dist_thresh=dist_thresh)
```

```python
## Get proportion of overlapping cells
prop_df = get_overlap_proportions(overlap_df, num_cells, markers, denom_marker)
prop_df
```

```python
## Save results
num_cells.to_csv(pjoin(output_path, 'number_of_marker_positive_cells.csv'))
overlap_df.to_csv(pjoin(output_path, 'overlapping_markers.csv'))
prop_df.to_csv(pjoin(output_path, 'marker_proportions.csv'))
```
